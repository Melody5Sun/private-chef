# 帮助文件：Delete
## 这是一个可以清除db文件的程序，简单来说，你在main_code中与智能体的对话都会被存在resources文件夹内的数据库文件里，运行这个代码块可以删除这个数据库，方便你再次测试。

```python
import os

# 数据库文件路径（如果resources和main_code在同一个文件夹，这个路径会自动找到db文件）
db_file_path = "resources/checkpoint.db"

if os.path.exists(db_file_path):
    os.remove(db_file_path)
    print(f"已删除数据库文件：{db_file_path}")
else:
    print("数据库文件不存在，无需删除。")
```
